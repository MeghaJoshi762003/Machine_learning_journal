# Univariate Analysis
## ->Categorical data  

```python
import seaborn as sns
```

```python
# countplot
sns.countplot(x='Gender',data=df)   #it will show the frequency f category
```

```python
df['Gender'].value_counts().plot(kind='bar')  #work same as above
```

```python
#pie chart
df['City'].value_counts().plot(kind='pie')
```

```python
df['Category'].value_counts().plot(kind='pie',autopct="%.2f")
```

```python
df['Payment_Mode'].value_counts().plot(kind='pie')
```
