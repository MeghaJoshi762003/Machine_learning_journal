## Numerical - Numerical

-> Scatterplot

```python
import seaborn as sns
sns.scatterplot(x=df['Support_Tickets_Raised'],y=df['Reminder_Emails_Clicked'],hue=df['Completed'],style=df['Fee_Paid'],size=df['Discount_Used'])
```

```python
df.select_dtypes(include='int').dtypes
```

```python
sns.scatterplot(x=df['Course_Duration_Days'],y=df['Support_Tickets_Raised'])
```

PairPlot   {give scatter plot between every numerical attribute}

```python
sns.pairplot(df)
```

LinePlot   {when we have any time based attribute like date,year,time}

```python
new=df.groupby('Enrollment_Date').sum().reset_index()
```

```python
sns.lineplot(x=new['Enrollment_Date'],y=new['Reminder_Emails_Clicked'])
```

->  Numerical-Categorical
BarPlot

```python
sns.barplot(x=df['Completed'],y=df['Payment_Amount'])
```

```python
sns.barplot(x=df['Completed'],y=df['Assignments_Submitted'])
```

```python
sns.barplot(x=df['Completed'],y=df['Assignments_Submitted'],hue=df['Category'])
```

BoxPlot

```python
sns.boxplot(x=df['Gender'],y=df['Age'])
```

```python
sns.boxplot(x=df['Gender'],y=df['Age'],hue=df['Completed'])
```

```python
df.dtypes
```

Distplot

```python
sns.distplot(x=df[df['Completed']=='Completed']['App_Usage_Percentage'])
sns.distplot(x=df[df['Completed']=='Not Completed']['App_Usage_Percentage'])
```

-> Categorical-Categorical
Heatmap

```python
pd.crosstab(df['Category'],df['Completed'])
```

```python
sns.heatmap(pd.crosstab(df['Category'],df['Completed']))
```

Cluster Map   {it shows the closeness between different category}

```python
sns.clustermap(pd.crosstab(df['Fee_Paid'],df['Completed']))
```
